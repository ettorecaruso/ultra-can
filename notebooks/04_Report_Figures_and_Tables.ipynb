## 4. Archive results and figures (Colab only)

Saves `results/` + `figures/` as one timestamped zip on Drive. Skipped when Drive is not mounted.

```python
import os, shutil, subprocess, time
OUT = "/content/drive/MyDrive/ultra_can"
if os.path.isdir("/content/drive/MyDrive"):
    os.makedirs(OUT, exist_ok=True)
    archive = os.path.join(OUT, "ultra_can_results_%s.zip" % time.strftime("%Y-%m-%d_%H%M"))
    subprocess.run(["zip", "-r", "-q", archive, "results", "figures"], check=True)
    print("archived:", archive)
else:
    print("Drive not mounted: skipping the archive")
```
