```python
import os, subprocess
REPO = "/content/ultra-can"
URL = "https://github.com/ettorecaruso/ultra-can.git"
if not os.path.isdir(os.path.join(REPO, ".git")):
    if not os.path.isdir(REPO):
        subprocess.run(["git", "clone", "-q", URL, REPO], check=True)
else:
    subprocess.run(["git", "-C", REPO, "pull", "-q", "--ff-only"], check=True)
if os.path.isdir(os.path.join(REPO, ".git")):
    os.chdir(REPO)
commit = subprocess.run(["git", "-C", os.getcwd(), "rev-parse", "--short", "HEAD"],
                        capture_output=True, text=True).stdout.strip()
print("Repo:", os.getcwd(), "| commit:", commit)
```
